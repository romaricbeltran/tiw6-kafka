```python
import pandas as pd
import re
import numpy as np
import requests
import urllib.request
import time
from datetime import date
import json
import wget
import gzip
import os
import re
import shutil
```

```python
y=[18,19]
m=list(range(1,13))
for year in y:
    for month in m:
        if month <10:
            wget.download('http://192.168.76.159/data/meteo/synop/synop.20'+str(year)+'0'+str(month)+'.csv.gz', out='../hist_data_meteo_france/')
        else:
            wget.download('http://192.168.76.159/data/meteo/synop/synop.20'+str(year)+str(month)+'.csv.gz', out='../hist_data_meteo_france/')
```

```python
files = os.listdir('../hist_data_meteo_france/')
files[1]
```

```python
reg=re.compile('synop\.([0-9][0-9][0-9][0-9][0-9][0-9])\.',re.IGNORECASE)
```

```python
print(re.search(reg,files[2]))
```

```python
for file in files:
    if re.search(reg,file):
        with gzip.open('../hist_data_meteo_france/'+file) as gz:
            with open('../data_meteo_france/'+file[:16], 'wb') as f_out:
                shutil.copyfileobj(gz, f_out)
```

```python
with gzip.open('../hist_data_meteo_france/synop.201912.csv.gz') as gz:
    with open('../data_meteo_france/synop.201912.csv', 'wb') as f_out:
        shutil.copyfileobj(gz, f_out)
```
